```python
import csv
import os

DEPTHS = [3, 5, 7, 10]
BRANCHING_FACTOR = 3
OUTPUT_DIR = "C:/Users/USER/OneDrive - Bina Nusantara/Documents/Semester 4/Research Methology"

ROLES = [
    "CEO", "Vice President", "Director", "Senior Manager",
    "Manager", "Senior Staff", "Staff", "Associate",
    "Junior Associate", "Intern",
]

NAMES = [
    "Alice", "Bob", "Carol", "Dave", "Eve", "Frank", "Grace", "Hank",
    "Iris", "Jack", "Karen", "Leo", "Maya", "Nate", "Olivia", "Paul",
    "Quinn", "Rose", "Sam", "Tina", "Uma", "Victor", "Wendy", "Xander",
    "Yara", "Zack", "Ava", "Ben", "Chloe", "Dylan", "Ella", "Finn",
]

def get_name(node_id):
    return NAMES[(node_id - 1) % len(NAMES)] + f"_{node_id}"

def generate_hierarchy(depth, branching):
    nodes = []
    node_id = 1
    current_level_ids = [node_id]
    nodes.append((node_id, get_name(node_id), ROLES[0], None))
    for level in range(1, depth):
        next_level_ids = []
        role = ROLES[level] if level < len(ROLES) else f"Level_{level+1}"
        for parent_id in current_level_ids:
            for _ in range(branching):
                node_id += 1
                nodes.append((node_id, get_name(node_id), role, parent_id))
                next_level_ids.append(node_id)
        current_level_ids = next_level_ids
    return nodes

def save_relational(nodes, depth):
    path = os.path.join(OUTPUT_DIR, f"employees_depth{depth}.csv")
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["id", "name", "role", "parent_id"])
        for (nid, name, role, parent_id) in nodes:
            writer.writerow([nid, name, role, parent_id if parent_id is not None else ""])
    print(f"  [relational] {path} ({len(nodes)} rows)")

def save_graph(nodes, depth):
    nodes_path = os.path.join(OUTPUT_DIR, f"nodes_depth{depth}.csv")
    rels_path  = os.path.join(OUTPUT_DIR, f"relationships_depth{depth}.csv")
    with open(nodes_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["id", "name", "role"])
        for (nid, name, role, _) in nodes:
            writer.writerow([nid, name, role])
    with open(rels_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["source_id", "target_id", "type"])
        for (nid, _, _, parent_id) in nodes:
            if parent_id is not None:
                writer.writerow([nid, parent_id, "REPORTS_TO"])
    print(f"  [graph]      {nodes_path} ({len(nodes)} nodes)")
    print(f"  [graph]      {rels_path} ({len(nodes)-1} relationships)")

os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"Generating datasets | branching={BRANCHING_FACTOR} | depths={DEPTHS}\n")
for depth in DEPTHS:
    nodes = generate_hierarchy(depth, BRANCHING_FACTOR)
    print(f"Depth {depth}: {len(nodes)} total nodes")
    save_relational(nodes, depth)
    save_graph(nodes, depth)
    print()
print("Done! Files saved to:", OUTPUT_DIR)
```
